# Step 3: Calculate 101 Formulaic Alphas for the Universe

This notebook computes the 101 formulaic alphas across all stocks in our tradable universe. It utilizes the `calculate_universe_alphas_parallel` function which handles:
- Parallel processing over multiple tickers.
- **Delay-0 and Delay-1 Shifts**: Prevents look-ahead bias by aligning signals properly.
- **Universe Masking**: Filters out untradable stocks to save memory and ignore noisy data.

```python
import os
import sys

# Add phase2_qrt_challenge to the Python path so we can import the scripts
sys.path.insert(0, os.path.abspath("phase2_qrt_challenge"))

from scripts.calculate_universe_alphas import calculate_universe_alphas_parallel
```

```python
# Define input and output paths relative to the notebook directory
YF_DATA_PATH = "top_5000_yf_data.pkl"
RETURNS_PATH = "stores/returns.parquet"
UNIVERSE_PATH = "stores/universe_5m.parquet"
OUTPUT_PATH = "stores/universe_alpha_factors.parquet"

# Execute the parallel alpha calculation
# This will utilize multiple CPU cores and may take a few minutes
calculate_universe_alphas_parallel(
    yf_data_path=YF_DATA_PATH,
    returns_path=RETURNS_PATH,
    universe_path=UNIVERSE_PATH,
    output_path=OUTPUT_PATH,
    n_jobs=4,  # Adjust the number of workers based on your machine's CPU
    chunk_size=50
)
```

```python
import pandas as pd

# Load the generated alpha factors to inspect
print("Loading computed alphas...")
alphas_df = pd.read_parquet(OUTPUT_PATH)

print(f"Shape of computed alphas: {alphas_df.shape}")
alphas_df.head()
```
